Image Point Processing Techniques
**Negative transformation** is useful when important structures are difficult to observe in the dark regions.

Black regions become white.

White regions become black.

Details present in dark regions may become easier to see.

```python
image_path = r"sample images\sample images\\"
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the X-ray image
img = cv2.imread(
    image_path + "Xray.png",
    cv2.IMREAD_GRAYSCALE
)

# Check if the image was loaded successfully
if img is None:
    print("Error: Could not load Xray.png")
else:
    # Negative transformation
    negative = 255 - img

    # Display
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Original X-ray")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(negative, cmap="gray")
    plt.title("Negative Transformation")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

# **Log Transform**
Dark pixels are expanded.

Bright pixels are compressed.

Hidden details in dark regions become more visible.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load mountain4.jpg using Pillow
img = np.array(
    Image.open(r"sample images\sample images\mountain4.jpg").convert("L")
)

print("Image loaded:", img is not None)
print("Image shape:", img.shape)

# Log transformation
img_float = img.astype(np.float32)

c = 255 / np.log(1 + np.max(img_float))
log_image = c * np.log(1 + img_float)

log_image = np.uint8(log_image)

# Display
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Dark Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_image, cmap="gray")
plt.title("Log Transformation")
plt.axis("off")

plt.tight_layout()
plt.show()
```
