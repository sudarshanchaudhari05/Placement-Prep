*Gamma Correction

## **Gamma correction adjusts image brightness non-linearly**.

**For dark images:**

Use γ < 1 (e.g., 0.5).

Expands the intensity values of dark pixels.

Makes dark regions brighter.

Reveals details that are difficult to see in shadows.

**For bright images:**

Use γ > 1 (e.g., 2.0).

Compresses high-intensity pixel values.

Makes the image darker.

Helps reveal details in overly bright regions.

**γ = 1: No change to the image.**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(
    r"sample images\sample images\mountain4.jpg",
    cv2.IMREAD_GRAYSCALE
)

# Check whether OpenCV can read the image
print("OpenCV loaded:", img is not None)
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load mountain4.jpg using Pillow
img = np.array(
    Image.open(r"sample images\sample images\mountain4.jpg").convert("L")
)

print("Image loaded:", img is not None)
print("Image shape:", img.shape)
```

```python
# Normalize pixel values to 0–1
img_normalized = img / 255.0

# Gamma values
gamma_values = [0.5, 1.0, 2.0]

plt.figure(figsize=(15, 4))

for i, gamma in enumerate(gamma_values):
    gamma_corrected = np.power(img_normalized, gamma)
    gamma_corrected = np.uint8(gamma_corrected * 255)

    plt.subplot(1, 3, i + 1)
    plt.imshow(gamma_corrected, cmap="gray")
    plt.title(f"Gamma = {gamma}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Contrast Stretching**

```python
# Read low contrast image
img = cv2.imread(image_path + "lowcontrast4.jpg", cv2.IMREAD_GRAYSCALE)


# Find minimum and maximum intensity
r_min = np.min(img)
r_max = np.max(img)

print("Minimum intensity:", r_min)
print("Maximum intensity:", r_max)

# Original histogram

plt.hist(img.ravel(), bins=256, range=(0, 256))
plt.title("Histogram - Original")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.xlim(0, 255)
```

```python
# Choose the range to be stretched

r_min = 100
r_max = 225

# Contrast Stretching Formula: s = ((r - r_min) / (r_max - r_min)) * 255
stretched = ((img.astype(np.float32) - r_min) /
             (r_max - r_min)) * 255

# Convert to uint8
stretched = np.clip(stretched, 0, 255).astype(np.uint8)

# Display Images and Histograms

plt.figure(figsize=(12, 8))

# Original image
plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

# Contrast stretched image
plt.subplot(2, 2, 2)
plt.imshow(stretched, cmap="gray")
plt.title("Contrast Stretched Image")
plt.axis("off")

# Original histogram
plt.subplot(2, 2, 3)
plt.hist(img.ravel(), bins=256, range=(0, 256))
plt.title("Histogram - Original")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.xlim(0, 255)

# Stretched histogram
plt.subplot(2, 2, 4)
plt.hist(stretched.ravel(), bins=256, range=(0, 256))
plt.title("Histogram - Contrast Stretched")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.xlim(0, 255)

plt.tight_layout()
plt.show()
```

**Bilateral Filter**

1. Reduces noise while preserving edges.
2. Prevents excessive blurring of object boundaries.
3. Useful as a preprocessing step before edge detection and segmentation.

**Bilateral Filter important parameters**

d → Size of the neighborhood.

sigmaColor → Controls how different pixel intensities can be mixed.

sigmaSpace → Controls how far spatially neighboring pixels can influence each other.




```python
# Bilateral Filter

# Input image: Gaussian noisy image
input_image = gaussian_image

# Combination 1
bilateral_1 = cv2.bilateralFilter(
    input_image,
    d=5,
    sigmaColor=30,
    sigmaSpace=30
)

# Combination 2
bilateral_2 = cv2.bilateralFilter(
    input_image,
    d=9,
    sigmaColor=75,
    sigmaSpace=75
)

# Combination 3
bilateral_3 = cv2.bilateralFilter(
    input_image,
    d=15,
    sigmaColor=150,
    sigmaSpace=150
)

# Display Input and Outputs

plt.figure(figsize=(14, 8))

# Input image
plt.subplot(2, 2, 1)
plt.imshow(input_image)
plt.title("Input: Gaussian Noisy Image")
plt.axis("off")

# Combination 1
plt.subplot(2, 2, 2)
plt.imshow(bilateral_1)
plt.title("d=5, σColor=30, σSpace=30")
plt.axis("off")

# Combination 2
plt.subplot(2, 2, 3)
plt.imshow(bilateral_2)
plt.title("d=9, σColor=75, σSpace=75")
plt.axis("off")

# Combination 3
plt.subplot(2, 2, 4)
plt.imshow(bilateral_3)
plt.title("d=15, σColor=150, σSpace=150")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **CLAHE ( Contrast Limited Adaptive Histogram Equalization)**

clipLimit → Controls contrast amplification and limits excessive noise amplification.

tileGridSize → Divides the image into small regions/tiles for local histogram equalization.



```python
# CLAHE - Contrast Limited Adaptive Histogram Equalization

# Read Lena image
image = cv2.imread(image_path + "flower2.jpg")

# Convert BGR to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


# CLAHE - Different Parameter Combinations

# Combination 1
clahe_1 = cv2.createCLAHE(
    clipLimit=1.0,
    tileGridSize=(6, 6)
)

clahe_img_1 = clahe_1.apply(gray)


# Combination 2
clahe_2 = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(6, 6)
)

clahe_img_2 = clahe_2.apply(gray)


# Combination 3
clahe_3 = cv2.createCLAHE(
    clipLimit=4.0,
    tileGridSize=(6, 6)
)

clahe_img_3 = clahe_3.apply(gray)

# Display Input and CLAHE Outputs

plt.figure(figsize=(10, 8))

# Input
plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Input: Lena Grayscale")
plt.axis("off")

# CLAHE 1
plt.subplot(2, 2, 2)
plt.imshow(clahe_img_1, cmap="gray")
plt.title("CLAHE: clipLimit=1.0, Grid=(6,6)")
plt.axis("off")

# CLAHE 2
plt.subplot(2, 2, 3)
plt.imshow(clahe_img_2, cmap="gray")
plt.title("CLAHE: clipLimit=2.0, Grid=(6,6)")
plt.axis("off")

# CLAHE 3
plt.subplot(2, 2, 4)
plt.imshow(clahe_img_3, cmap="gray")
plt.title("CLAHE: clipLimit=4.0, Grid=(6,6)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Lena image

image = cv2.imread(image_path + "lena.bmp")

# Convert BGR to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


# 1. Canny Edge Detection

canny = cv2.Canny(gray, 100, 200)


# 2. Sobel Edge Detection

# Sobel X and Y
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

# Calculate gradient magnitude
sobel = np.sqrt(sobel_x**2 + sobel_y**2)

# Convert to uint8
sobel = np.uint8(np.clip(sobel, 0, 255))


# 3. Prewitt Edge Detection

# Prewitt kernels
prewitt_x = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

prewitt_y = np.array([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
])

# Apply kernels
prewitt_x_result = cv2.filter2D(
    gray, cv2.CV_64F, prewitt_x
)

prewitt_y_result = cv2.filter2D(
    gray, cv2.CV_64F, prewitt_y
)

# Gradient magnitude
prewitt = np.sqrt(
    prewitt_x_result**2 +
    prewitt_y_result**2
)

# Convert to uint8
prewitt = np.uint8(
    np.clip(prewitt, 0, 255)
)


# Display Input and All Edge Detectors

plt.figure(figsize=(14, 8))

# Input
plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Input: Lena Grayscale")
plt.axis("off")

# Canny
plt.subplot(2, 2, 2)
plt.imshow(canny, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

# Sobel
plt.subplot(2, 2, 3)
plt.imshow(sobel, cmap="gray")
plt.title("Sobel Edge Detection")
plt.axis("off")

# Prewitt
plt.subplot(2, 2, 4)
plt.imshow(prewitt, cmap="gray")
plt.title("Prewitt Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Morphological Operations**
Morphological operations are image processing techniques based on the shape and structure of objects. They use a structuring element (kernel) to modify binary or grayscale images.

They are mainly used to
remove noise,
fill gaps,
separate or connect objects,
extract boundaries, and
improve object shapes before further image analysis.

**Common operations**:

**Erosion** – shrinks objects

**Dilation** – expands objects

**Opening** – removes small noise

**Closing** – fills small gaps/holes

**Gradient** – extracts object boundaries

**Top Hat** – extracts small bright regions

**Black Hat** – extracts small dark regions



```python
# 1. Read Coins Image


image = cv2.imread(image_path + "coins.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. Thresholding

threshold = 120

_, binary = cv2.threshold(
    gray,
    threshold,
    255,
    cv2.THRESH_BINARY
)

# 3. Create Structuring Element
kernel = np.ones((5, 5), np.uint8)

print("Structuring Element:")
print(kernel)

# 4. Morphological Operations
# Erosion
erosion = cv2.erode(
    binary,
    kernel,
    iterations=1
)

# Dilation
dilation = cv2.dilate(
    binary,
    kernel,
    iterations=1
)

# Opening
opening = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    kernel
)

# Closing
closing = cv2.morphologyEx(
    binary,
    cv2.MORPH_CLOSE,
    kernel
)

# Morphological Gradient
gradient = cv2.morphologyEx(
    binary,
    cv2.MORPH_GRADIENT,
    kernel
)

# Top Hat
top_hat = cv2.morphologyEx(
    binary,
    cv2.MORPH_TOPHAT,
    kernel
)

# Black Hat
black_hat = cv2.morphologyEx(
    binary,
    cv2.MORPH_BLACKHAT,
    kernel
)


# 5. Display All Operations

plt.figure(figsize=(15, 12))

# Input
plt.subplot(3, 3, 1)
plt.imshow(binary, cmap="gray")
plt.title("Input: Binary Image")
plt.axis("off")

# Erosion
plt.subplot(3, 3, 2)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion")
plt.axis("off")

# Dilation
plt.subplot(3, 3, 3)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation")
plt.axis("off")

# Opening
plt.subplot(3, 3, 4)
plt.imshow(opening, cmap="gray")
plt.title("Opening")
plt.axis("off")

# Closing
plt.subplot(3, 3, 5)
plt.imshow(closing, cmap="gray")
plt.title("Closing")
plt.axis("off")

# Gradient
plt.subplot(3, 3, 6)
plt.imshow(gradient, cmap="gray")
plt.title("Morphological Gradient")
plt.axis("off")

# Top Hat
plt.subplot(3, 3, 7)
plt.imshow(top_hat, cmap="gray")
plt.title("Top Hat")
plt.axis("off")

# Black Hat
plt.subplot(3, 3, 8)
plt.imshow(black_hat, cmap="gray")
plt.title("Black Hat")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# **Data Augmentation**

Image augmentation is the process of creating modified versions of existing images by applying transformations such as rotation, flipping, scaling, cropping, translation, brightness/contrast adjustment, noise addition, and blurring.

**Why is Image Augmentation Needed?**

Increases the size of the training dataset without collecting new images.

Improves model generalization by exposing the model to different image variations.

Reduces overfitting by preventing the model from memorizing the training images.

Makes the model more robust to changes in position, orientation, scale, illumination, and image quality.

Helps handle limited or imbalanced datasets.

Improves performance when real-world images differ from the training images.


```python
import os
import cv2

apple_path = image_path + "apple.jpg"

print("Path:", os.path.abspath(apple_path))
print("File exists:", os.path.isfile(apple_path))

test = cv2.imread(apple_path)

print("OpenCV loaded:", test is not None)
```

```python

# 1. Read Image

from PIL import Image

image = np.array(
    Image.open(image_path + "apple.jpg").convert("RGB")
)

# Convert BGR to RGB
# Not needed because Pillow already gives RGB

print("Input Image Shape:", image.shape)

# Resize image to 256 × 256
image = cv2.resize(image, (256, 256))

# 2. Rotation
height, width = image.shape[:2]

center = (width // 2, height // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    30,       # angle
    1.0       # scale
)

rotated = cv2.warpAffine(
    image,
    rotation_matrix,
    (width, height)
)

# 3. Horizontal Flip
horizontal_flip = cv2.flip(image, 1)


# 4. Vertical Flip
vertical_flip = cv2.flip(image, 0)


# 5. Translation
tx = 100     # Shift right
ty = 50     # Shift down

translation_matrix = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

translated = cv2.warpAffine(
    image,
    translation_matrix,
    (width, height),
    borderMode=cv2.BORDER_CONSTANT,
    borderValue=(255, 255, 255)
)

# 6. Scaling
scale = 1.5

scaled = cv2.resize(
    image,
    None,
    fx=scale,
    fy=scale,
    interpolation=cv2.INTER_LINEAR
)

# Crop the center portion to maintain display size
scaled_height, scaled_width = scaled.shape[:2]

start_x = (scaled_width - width) // 2
start_y = (scaled_height - height) // 2

scaled = scaled[
    start_y:start_y + height,
    start_x:start_x + width
]


# 7. Cropping
crop = image[
    50:height-50,
    50:width-50
]

crop = cv2.resize(
    crop,
    (width, height)
)
print("Cropped Shape:", crop.shape)


# 8. Brightness Adjustment
brightness = cv2.convertScaleAbs( image, alpha=1.0, beta=40 )

# 9. Contrast Adjustment
contrast = cv2.convertScaleAbs(
    image,
    alpha=1.5,
    beta=0
)


# 10. Gaussian Noise
noise = np.random.normal(
    0,
    20,
    image.shape
)

noisy = image.astype(np.float32) + noise

noisy = np.clip(
    noisy,
    0,
    255
).astype(np.uint8)


# 11. Gaussian Blur
blurred = cv2.GaussianBlur(
    image,
    (5, 5),
    0
)


# 12. Display All Augmentations
plt.figure(figsize=(15, 12))

# Original
plt.subplot(3, 4, 1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

# Rotation
plt.subplot(3, 4, 2)
plt.imshow(rotated)
plt.title("Rotation: 30°")
plt.axis("off")

# Horizontal Flip
plt.subplot(3, 4, 3)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

# Vertical Flip
plt.subplot(3, 4, 4)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

# Translation
plt.subplot(3, 4, 5)
plt.imshow(translated)
plt.title("Translation")
plt.axis("off")

# Scaling
plt.subplot(3, 4, 6)
plt.imshow(scaled)
plt.title("Scaling")
plt.axis("off")

# Cropping
plt.subplot(3, 4, 7)
plt.imshow(crop)
plt.title("Cropping")
plt.axis("off")

# Brightness
plt.subplot(3, 4, 8)
plt.imshow(brightness)
plt.title("Brightness +40")
plt.axis("off")

# Contrast
plt.subplot(3, 4, 9)
plt.imshow(contrast)
plt.title("Contrast ×1.5")
plt.axis("off")

# Gaussian Noise
plt.subplot(3, 4, 10)
plt.imshow(noisy)
plt.title("Gaussian Noise")
plt.axis("off")

# Blur
plt.subplot(3, 4, 11)
plt.imshow(blurred)
plt.title("Gaussian Blur")
plt.axis("off")

plt.tight_layout()
plt.show()
```
